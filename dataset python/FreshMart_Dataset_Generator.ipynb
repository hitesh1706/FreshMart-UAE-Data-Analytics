```python
# ==========================================
# FreshMart UAE - Product Generator
# Cell 1 : Imports & Business Rules
# ==========================================

import pandas as pd
import random
from faker import Faker
from datetime import datetime

fake = Faker()

TARGET_PRODUCTS = 900

ACTIVE_RATIO = 0.95

EXPIRY_SUBCATEGORIES = {
    "Fresh Fruits",
    "Fresh Vegetables",
    "Herbs",
    "Milk",
    "Cheese",
    "Yogurt",
    "Butter",
    "Eggs",
    "Bread",
    "Cakes",
    "Pastries",
    "Water",
    "Juices",
    "Frozen Vegetables",
    "Ice Cream",
    "Frozen Chicken",
    "Chicken",
    "Beef",
    "Fish",
    "Seafood",
    "Baby Food"
}

price_ranges = {
    "Fresh Fruits": (2, 25),
    "Fresh Vegetables": (2, 20),
    "Herbs": (1, 8),
    "Milk": (4, 18),
    "Cheese": (8, 45),
    "Yogurt": (3, 20),
    "Butter": (6, 30),
    "Eggs": (8, 35),
    "Bread": (3, 15),
    "Cakes": (20, 120),
    "Pastries": (4, 18),
    "Water": (1, 12),
    "Soft Drinks": (3, 15),
    "Juices": (5, 22),
    "Tea": (8, 40),
    "Coffee": (15, 60),
    "Energy Drinks": (8, 20),
    "Chips": (3, 18),
    "Biscuits": (4, 22),
    "Chocolates": (5, 30),
    "Nuts": (10, 60),
    "Frozen Vegetables": (8, 25),
    "Ice Cream": (12, 45),
    "Frozen Chicken": (15, 40),
    "Chicken": (12, 45),
    "Beef": (20, 80),
    "Fish": (18, 70),
    "Seafood": (25, 120),
    "Rice": (10, 60),
    "Pasta": (4, 20),
    "Flour": (5, 25),
    "Sugar": (4, 20),
    "Cooking Oil": (12, 60),
    "Spices": (3, 25),
    "Shampoo": (12, 45),
    "Soap": (3, 20),
    "Toothpaste": (6, 25),
    "Skincare": (15, 80),
    "Tissue": (4, 20),
    "Garbage Bags": (6, 18),
    "Foil": (5, 20),
    "Storage Bags": (5, 20),
    "Detergent": (15, 60),
    "Floor Cleaner": (10, 35),
    "Dishwashing Liquid": (8, 30),
    "Diapers": (35, 120),
    "Baby Food": (10, 40),
    "Baby Wipes": (8, 30)
}
```

```python
# ==========================================
# FreshMart UAE - Product Generator
# Cell 2 : Generate Products
# ==========================================

products = []
product_names = set()
product_counter = 1

while len(products) < TARGET_PRODUCTS:

    # Random Category
    category = random.choice(list(product_templates.keys()))

    # Random Sub Category
    sub_category = random.choice(list(product_templates[category].keys()))

    # Random Template
    template = random.choice(product_templates[category][sub_category])

    # Valid Brand
    brand = random.choice(brand_master[sub_category])

    # Valid Size
    size = random.choice(package_sizes[sub_category])

    # Final Product Name
    product_name = f"{brand} {template} {size}"

    # Avoid duplicates
    if product_name in product_names:
        continue

    product_names.add(product_name)

    # Product ID
    product_id = f"P{product_counter:04d}"

    # Unit Cost (60-80% of Selling Price)
    min_price, max_price = price_ranges[sub_category]

    selling_price = round(random.uniform(min_price, max_price), 2)

    cost_percentage = random.uniform(0.60, 0.80)

    unit_cost = round(selling_price * cost_percentage, 2)

    profit_margin = round(
        ((selling_price - unit_cost) / selling_price) * 100,
        2
    )

    # Barcode
    barcode = fake.unique.ean13()

    # Expiry
    expiry_required = (
        "Yes"
        if sub_category in EXPIRY_SUBCATEGORIES
        else "No"
    )

    # Launch Date
    launch_date = fake.date_between(
        start_date="-5y",
        end_date="today"
    )

    # Status
    status = random.choices(
        ["Active", "Discontinued"],
        weights=[95, 5]
    )[0]

    products.append({
        "product_id": product_id,
        "product_name": product_name,
        "category": category,
        "sub_category": sub_category,
        "brand": brand,
        "size": size,
        "unit_cost": unit_cost,
        "selling_price": selling_price,
        "profit_margin": profit_margin,
        "barcode": barcode,
        "expiry_required": expiry_required,
        "launch_date": launch_date,
        "status": status
    })

    product_counter += 1

products_df = pd.DataFrame(products)
products_df.to_csv(
    "products.csv",
    index=False
)

print("products.csv created successfully")
print("Total Products:", len(products_df))
```

```python
# ==========================================
# FreshMart UAE - Supplier Generator
# ==========================================

NUM_SUPPLIERS = 100

import pandas as pd
import random
from faker import Faker

fake = Faker()

suppliers = []

supplier_counter = 1


# UAE Locations

supplier_locations = {
    "Dubai": "Dubai",
    "Abu Dhabi": "Abu Dhabi",
    "Sharjah": "Sharjah",
    "Ajman": "Ajman",
    "Ras Al Khaimah": "Ras Al Khaimah",
    "Fujairah": "Fujairah",
    "Umm Al Quwain": "Umm Al Quwain",
    "Al Ain": "Abu Dhabi"
}


# Supplier naming variations

supplier_suffixes = [
    "Distribution Division",
    "Wholesale Partner",
    "Trading LLC",
    "UAE Branch",
    "Retail Supply Division",
    "Food Trading",
    "Commercial Division",
    "Regional Supplier"
]


existing_names = set()


while len(suppliers) < NUM_SUPPLIERS:

    # Select category type
    supplier_type_key = random.choice(
        list(supplier_categories.keys())
    )

    # Select actual category
    category = random.choice(
        supplier_categories[supplier_type_key]
    )

    # Select base supplier name
    base_name = random.choice(
        supplier_master[category]
    )


    # Create unique supplier name

    supplier_name = (
        base_name
        + " - "
        + random.choice(supplier_suffixes)
    )


    # Avoid duplicates

    if supplier_name in existing_names:
        continue


    existing_names.add(supplier_name)


    city = random.choice(
        list(supplier_locations.keys())
    )


    supplier_id = f"S{supplier_counter:04d}"


    suppliers.append({

        "supplier_id": supplier_id,

        "supplier_name": supplier_name,

        "supplier_type": random.choice(
            [
                "Local",
                "International"
            ]
        ),

        "category": category,

        "city": city,

        "emirate": supplier_locations[city],

        "contact_person": fake.name(),

        "email": fake.company_email(),

        "phone": fake.phone_number(),

        "contract_start_date": fake.date_between(
            start_date="-10y",
            end_date="-1y"
        ),

        "payment_terms": random.choice(
            [
                "30 Days",
                "45 Days",
                "60 Days"
            ]
        ),

        "status": random.choice(
            [
                "Active",
                "Inactive"
            ]
        )
    })


    supplier_counter += 1



suppliers_df = pd.DataFrame(suppliers)


suppliers_df.to_csv(
    "suppliers.csv",
    index=False
)


print("suppliers.csv created successfully")
print("Total Suppliers:", len(suppliers_df))
```

```python
# ==========================================
# FreshMart UAE - Order Generator
# ==========================================

import pandas as pd
import random
from faker import Faker
from datetime import timedelta

fake = Faker()

customers = pd.read_csv("customers.csv")
stores = pd.read_csv("stores.csv")

orders = []

status_list = list(ORDER_STATUS.keys())
status_weights = list(ORDER_STATUS.values())

payment_list = list(PAYMENT_METHODS.keys())
payment_weights = list(PAYMENT_METHODS.values())

delivery_list = list(DELIVERY_TYPE.keys())
delivery_weights = list(DELIVERY_TYPE.values())

for order_num in range(1, NUM_ORDERS + 1):

    customer = customers.sample(1).iloc[0]
    store = stores.sample(1).iloc[0]

    order_date = fake.date_time_between(
        start_date="-2y",
        end_date="now"
    )

    order_status = random.choices(
        status_list,
        weights=status_weights,
        k=1
    )[0]

    delivery_type = random.choices(
        delivery_list,
        weights=delivery_weights,
        k=1
    )[0]

    payment_method = random.choices(
        payment_list,
        weights=payment_weights,
        k=1
    )[0]

    # Delivery Days

    if order_status == "Cancelled":
        delivery_date = None

    elif order_status == "Processing":
        delivery_date = None

    else:

        if delivery_type == "Express":
            days = random.randint(0, 1)

        elif delivery_type == "Standard":
            days = random.randint(2, 5)

        else:
            days = random.randint(0, 2)

        delivery_date = (
            order_date + timedelta(days=days)
        ).date()

    shipping_fee = {
        "Standard": 10,
        "Express": 25,
        "Pickup": 0
    }[delivery_type]

    orders.append({

        "order_id": f"O{order_num:06d}",

        "customer_id": customer["customer_id"],

        "store_id": store["store_id"],

        "order_date": order_date.date(),

        "delivery_date": delivery_date,

        "order_status": order_status,

        "delivery_type": delivery_type,

        "payment_method": payment_method,

        "shipping_fee": shipping_fee

    })

orders_df = pd.DataFrame(orders)

orders_df.to_csv(
    "orders.csv",
    index=False
)

print("orders.csv created successfully")
print("Total Orders:", len(orders_df))

orders_df.head()
```

```python
# ==========================================
# FreshMart UAE - Order Items Configuration
# ==========================================

MIN_ITEMS_PER_ORDER = 1
MAX_ITEMS_PER_ORDER = 5

QUANTITY_WEIGHTS = {
    1: 55,
    2: 25,
    3: 12,
    4: 5,
    5: 3
}

DISCOUNT_OPTIONS = {
    0: 40,
    5: 25,
    10: 20,
    15: 10,
    20: 5
}
```

```python
# ==========================================
# FreshMart UAE - Order Items Generator
# ==========================================

import pandas as pd
import random

orders = pd.read_csv("orders.csv")
products = pd.read_csv("products.csv")

order_items = []

item_counter = 1

quantity_values = list(QUANTITY_WEIGHTS.keys())
quantity_weights = list(QUANTITY_WEIGHTS.values())

discount_values = list(DISCOUNT_OPTIONS.keys())
discount_weights = list(DISCOUNT_OPTIONS.values())

for _, order in orders.iterrows():

    items_in_order = random.randint(
        MIN_ITEMS_PER_ORDER,
        MAX_ITEMS_PER_ORDER
    )

    selected_products = products.sample(
        n=items_in_order,
        replace=False
    )

    for _, product in selected_products.iterrows():

        quantity = random.choices(
            quantity_values,
            weights=quantity_weights,
            k=1
        )[0]

        unit_price = float(product["selling_price"])

        discount_percent = random.choices(
            discount_values,
            weights=discount_weights,
            k=1
        )[0]

        discount_amount = round(
            unit_price * discount_percent / 100,
            2
        )

        final_unit_price = round(
            unit_price - discount_amount,
            2
        )

        line_total = round(
            final_unit_price * quantity,
            2
        )

        order_items.append({

            "order_item_id": f"OI{item_counter:07d}",

            "order_id": order["order_id"],

            "product_id": product["product_id"],

            "quantity": quantity,

            "unit_price": unit_price,

            "discount_percent": discount_percent,

            "discount_amount": discount_amount,

            "final_unit_price": final_unit_price,

            "line_total": line_total

        })

        item_counter += 1

order_items_df = pd.DataFrame(order_items)

order_items_df.to_csv(
    "order_items.csv",
    index=False
)

print("order_items.csv created successfully")
print("Total Order Items:", len(order_items_df))

order_items_df.head()
```

```python
# ==========================================
# FreshMart UAE - Update Orders with Totals
# ==========================================

import pandas as pd

orders = pd.read_csv("orders.csv")
order_items = pd.read_csv("order_items.csv")

# Calculate totals for each order
order_summary = (
    order_items
    .groupby("order_id")
    .agg(
        total_items=("quantity", "sum"),
        total_discount=("discount_amount", "sum"),
        order_total=("line_total", "sum")
    )
    .reset_index()
)

# Merge into orders
orders = orders.merge(
    order_summary,
    on="order_id",
    how="left"
)

# Round values
orders["total_discount"] = orders["total_discount"].round(2)
orders["order_total"] = orders["order_total"].round(2)

# Save updated orders
orders.to_csv("orders.csv", index=False)

print("orders.csv updated successfully!")

orders.head()
```

```python
# ==========================================
# FreshMart UAE - Returns Generator
# ==========================================

import pandas as pd
import random

orders = pd.read_csv("orders.csv")

# Only returned orders
returned_orders = orders[
    orders["order_status"] == "Returned"
].copy()

return_reasons = [
    "Damaged Product",
    "Wrong Item",
    "Expired Product",
    "Poor Quality",
    "Customer Changed Mind",
    "Missing Items"
]

return_statuses = [
    "Approved",
    "Rejected",
    "Refund Completed"
]

returns = []

return_counter = 1

for _, order in returned_orders.iterrows():

    delivery_date = pd.to_datetime(order["delivery_date"])

    # Return created 1–15 days after delivery
    return_date = (
        delivery_date +
        pd.Timedelta(days=random.randint(1, 15))
    ).date()

    returns.append({

        "return_id": f"R{return_counter:06d}",

        "order_id": order["order_id"],

        "customer_id": order["customer_id"],

        "return_date": return_date,

        "return_reason": random.choice(return_reasons),

        "return_status": random.choices(
            return_statuses,
            weights=[40, 10, 50],
            k=1
        )[0]

    })

    return_counter += 1

returns_df = pd.DataFrame(returns)

returns_df.to_csv(
    "returns.csv",
    index=False
)

print("returns.csv created successfully")
print("Total Returns:", len(returns_df))

returns_df.head()
```

```python
# ==========================================
# FreshMart UAE - Promotions Generator
# ==========================================

import pandas as pd
import random
from faker import Faker

fake = Faker()

products = pd.read_csv("products.csv")

NUM_PROMOTIONS = 100

promotion_names = [
    "Weekend Sale",
    "Mega Discount",
    "Ramadan Offer",
    "Summer Sale",
    "Buy More Save More",
    "Family Pack Offer",
    "Flash Sale",
    "Festive Offer",
    "Loyalty Bonus",
    "New Arrival Offer"
]

promotions = []

for i in range(1, NUM_PROMOTIONS + 1):

    product = products.sample(1).iloc[0]

    start_date = fake.date_between(
        start_date="-2y",
        end_date="today"
    )

    duration = random.randint(7, 30)

    end_date = (
        pd.to_datetime(start_date) +
        pd.Timedelta(days=duration)
    ).date()

    discount = random.choice(
        [5, 10, 15, 20, 25, 30]
    )

    promotions.append({

        "promotion_id": f"PR{i:04d}",

        "promotion_name": random.choice(
            promotion_names
        ),

        "product_id": product["product_id"],

        "discount_percent": discount,

        "start_date": start_date,

        "end_date": end_date,

        "status": random.choices(
            ["Active", "Expired"],
            weights=[20, 80],
            k=1
        )[0]

    })

promotions_df = pd.DataFrame(promotions)

promotions_df.to_csv(
    "promotions.csv",
    index=False
)

print("promotions.csv created successfully")
print("Total Promotions:", len(promotions_df))

promotions_df.head()
```

```python
# ==========================================
# FreshMart UAE - Reviews Generator
# ==========================================

import pandas as pd
import random
from faker import Faker

fake = Faker()

orders = pd.read_csv("orders.csv")

# Only delivered orders
delivered_orders = orders[
    orders["order_status"] == "Delivered"
].copy()

# Around 40% of delivered orders leave a review
reviews_orders = delivered_orders.sample(
    frac=0.40,
    random_state=42
)

review_comments = {
    5: [
        "Excellent quality",
        "Highly recommended",
        "Very satisfied",
        "Fresh products",
        "Fast delivery"
    ],
    4: [
        "Good service",
        "Nice packaging",
        "Value for money",
        "Satisfied"
    ],
    3: [
        "Average experience",
        "Delivery was okay",
        "Could be better"
    ],
    2: [
        "Late delivery",
        "Packaging issue",
        "Product quality below expectations"
    ],
    1: [
        "Very disappointed",
        "Wrong item received",
        "Damaged product"
    ]
}

ratings = [5, 4, 3, 2, 1]
rating_weights = [55, 25, 10, 6, 4]

reviews = []

review_counter = 1

for _, order in reviews_orders.iterrows():

    rating = random.choices(
        ratings,
        weights=rating_weights,
        k=1
    )[0]

    review_date = (
        pd.to_datetime(order["delivery_date"])
        + pd.Timedelta(days=random.randint(1, 10))
    ).date()

    reviews.append({

        "review_id": f"RV{review_counter:06d}",

        "order_id": order["order_id"],

        "customer_id": order["customer_id"],

        "rating": rating,

        "review_comment": random.choice(
            review_comments[rating]
        ),

        "review_date": review_date

    })

    review_counter += 1

reviews_df = pd.DataFrame(reviews)

reviews_df.to_csv(
    "reviews.csv",
    index=False
)

print("reviews.csv created successfully")
print("Total Reviews:", len(reviews_df))

reviews_df.head()
```

```python
# ==========================================
# FreshMart UAE - QA Step 3
# Business Logic Validation
# ==========================================

import pandas as pd

orders = pd.read_csv("orders.csv")
payments = pd.read_csv("payments.csv")
returns = pd.read_csv("returns.csv")
reviews = pd.read_csv("reviews.csv")
inventory = pd.read_csv("inventory.csv")
promotions = pd.read_csv("promotions.csv")

print("=" * 60)
print("BUSINESS LOGIC VALIDATION")
print("=" * 60)

# -------------------------------------------------
# 1. Payment Amount Check
# -------------------------------------------------
merged = orders.merge(
    payments,
    on="order_id",
    suffixes=("_order", "_payment")
)

payment_check = (
    (merged["order_total"] + merged["shipping_fee"]).round(2)
    == merged["payment_amount"].round(2)
).all()

print("\n1. Payment Amount Validation:", payment_check)

# -------------------------------------------------
# 2. Returned Orders Check
# -------------------------------------------------
returned_orders = set(
    orders.loc[
        orders["order_status"] == "Returned",
        "order_id"
    ]
)

returns_check = set(
    returns["order_id"]
).issubset(returned_orders)

print("2. Returns Validation:", returns_check)

# -------------------------------------------------
# 3. Reviews Check
# -------------------------------------------------
delivered_orders = set(
    orders.loc[
        orders["order_status"] == "Delivered",
        "order_id"
    ]
)

reviews_check = set(
    reviews["order_id"]
).issubset(delivered_orders)

print("3. Reviews Validation:", reviews_check)

# -------------------------------------------------
# 4. Inventory Check
# -------------------------------------------------
inventory_check = (
    (
        (inventory["current_stock"] <= inventory["reorder_level"])
        ==
        (inventory["reorder_status"] == "Yes")
    )
).all()

print("4. Inventory Reorder Logic:", inventory_check)

# -------------------------------------------------
# 5. Promotion Dates
# -------------------------------------------------
promotions["start_date"] = pd.to_datetime(promotions["start_date"])
promotions["end_date"] = pd.to_datetime(promotions["end_date"])

promotion_check = (
    promotions["end_date"] >= promotions["start_date"]
).all()

print("5. Promotion Dates:", promotion_check)

print("\n" + "=" * 60)
print("QA COMPLETED")
print("=" * 60)
```

```python
# ==========================================
# FreshMart UAE - Fix Store Manager Mapping
# ==========================================

import pandas as pd

stores = pd.read_csv("stores.csv")
employees = pd.read_csv("employees.csv")

# Priority of roles
priority_roles = [
    "Store Manager",
    "Assistant Manager",
    "Supervisor"
]

manager_lookup = {}

# Find one manager for each store based on priority
for store in stores["store_id"]:

    selected_manager = None

    for role in priority_roles:

        emp = employees[
            (employees["store_id"] == store) &
            (employees["job_role"] == role)
        ]

        if not emp.empty:
            selected_manager = emp.iloc[0]["employee_id"]
            break

    manager_lookup[store] = selected_manager

# Update manager_id
stores["manager_id"] = stores["store_id"].map(manager_lookup)

# Save
stores.to_csv("stores.csv", index=False)

print("stores.csv updated successfully!")

stores.head()
```

```python
# ==========================================
# FreshMart UAE - Fix Customer Join Dates
# Compatible with Python 3.14 + Pandas
# ==========================================

import pandas as pd
import random

# Load files
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")

# Convert dates
customers["join_date"] = pd.to_datetime(customers["join_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])

# Get each customer's first order date
first_orders = (
    orders.groupby("customer_id", as_index=False)["order_date"]
    .min()
    .rename(columns={"order_date": "first_order_date"})
)

# Merge into customer table
customers = customers.merge(first_orders, on="customer_id", how="left")

# Find invalid customers
mask = (
    customers["first_order_date"].notna() &
    (customers["join_date"] > customers["first_order_date"])
)

print("Customers to fix:", mask.sum())

# Fix dates
for idx in customers[mask].index:

    first_order = customers.loc[idx, "first_order_date"]

    days_before = random.randint(30, 730)

    customers.loc[idx, "join_date"] = (
        first_order - pd.Timedelta(days=days_before)
    )

# Remove helper column
customers.drop(columns="first_order_date", inplace=True)

# Save
customers.to_csv("customers.csv", index=False)

print("customers.csv updated successfully!")
```

```python
# ==========================================
# FreshMart UAE - Fix Product Launch Dates
# Compatible with Python 3.14 + Pandas
# ==========================================

import pandas as pd
import random

# Load files
products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")
order_items = pd.read_csv("order_items.csv")

# Convert dates
products["launch_date"] = pd.to_datetime(products["launch_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])

# Find first sale date for every product
first_sales = (
    order_items
    .merge(
        orders[["order_id", "order_date"]],
        on="order_id",
        how="left"
    )
    .groupby("product_id", as_index=False)["order_date"]
    .min()
    .rename(columns={"order_date": "first_sale_date"})
)

# Merge with products
products = products.merge(first_sales, on="product_id", how="left")

# Find invalid products
mask = (
    products["first_sale_date"].notna() &
    (products["launch_date"] > products["first_sale_date"])
)

print("Products to fix:", mask.sum())

# Fix launch dates
for idx in products[mask].index:

    first_sale = products.loc[idx, "first_sale_date"]

    days_before = random.randint(30, 730)

    products.loc[idx, "launch_date"] = (
        first_sale - pd.Timedelta(days=days_before)
    )

# Remove helper column
products.drop(columns="first_sale_date", inplace=True)

# Save
products.to_csv("products.csv", index=False)

print("products.csv updated successfully!")
```

```python
# ==========================================
# FreshMart UAE - Fix Store Opening Dates
# Compatible with Python 3.14 + Pandas
# ==========================================

import pandas as pd
import random

# Load files
stores = pd.read_csv("stores.csv")
orders = pd.read_csv("orders.csv")

# Convert dates
stores["opening_date"] = pd.to_datetime(stores["opening_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])

# First order for each store
first_orders = (
    orders.groupby("store_id", as_index=False)["order_date"]
    .min()
    .rename(columns={"order_date": "first_order_date"})
)

# Merge
stores = stores.merge(first_orders, on="store_id", how="left")

# Find invalid stores
mask = (
    stores["first_order_date"].notna() &
    (stores["opening_date"] > stores["first_order_date"])
)

print("Stores to fix:", mask.sum())

# Fix dates
for idx in stores[mask].index:

    first_order = stores.loc[idx, "first_order_date"]

    days_before = random.randint(30, 730)

    stores.loc[idx, "opening_date"] = (
        first_order - pd.Timedelta(days=days_before)
    )

# Remove helper column
stores.drop(columns="first_order_date", inplace=True)

# Save
stores.to_csv("stores.csv", index=False)

print("stores.csv updated successfully!")
```
